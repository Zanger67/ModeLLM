```python
import os
```

```python
old_formatted_files = list(zip(*[os.listdir(os.path.join('exports', l)) for l in ['committee_histories', 'delegate_leaderboards', 'dialogue_transcripts', 'performance_metrics']]))
```

```python
new_dir = 'results'
for history, leaderboard, dialogue, metric in old_formatted_files:
    timestamp = history[-20:-5]
    
    os.makedirs(os.path.join(new_dir, timestamp), exist_ok=True)
    
    os.rename(os.path.join('exports', 'committee_histories', history), os.path.join(new_dir, timestamp, "committee_history.json"))
    os.rename(os.path.join('exports', 'delegate_leaderboards', leaderboard), os.path.join(new_dir, timestamp, "delegate_leaderboard.json"))
    os.rename(os.path.join('exports', 'dialogue_transcripts', dialogue), os.path.join(new_dir, timestamp, "dialogue_transcript.json"))
    os.rename(os.path.join('exports', 'performance_metrics', metric), os.path.join(new_dir, timestamp, "performance_metric.json"))
    
```
